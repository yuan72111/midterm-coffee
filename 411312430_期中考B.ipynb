````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（老師提供，請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
import sqlite3

def init_db():
    conn = sqlite3.connect('coffee_shop.db')
    cursor = conn.cursor()
    try:
        # 每次執行先刪除舊表，確保重置
        cursor.execute("DROP TABLE IF EXISTS orders")
        cursor.execute("DROP TABLE IF EXISTS menu")

        # 建立 menu 菜單表
        cursor.execute('''
            CREATE TABLE menu (
                item TEXT PRIMARY KEY,
                price INTEGER,
                stock INTEGER
            )
        ''')

        # 建立 orders 訂單表
        cursor.execute('''
            CREATE TABLE orders (
                order_id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT,
                item TEXT,
                quantity INTEGER,
                status TEXT
            )
        ''')

        # 放入初始資料
        initial_menu = [
            ('美式咖啡', 50, 5),
            ('拿鐵', 65, 4),
            ('抹茶歐蕾', 70, 3)
        ]
        cursor.executemany("INSERT INTO menu (item, price, stock) VALUES (?, ?, ?)", initial_menu)
        conn.commit()
    except Exception as e:
        print("初始化資料庫失敗：", e)
    finally:
        conn.close()

init_db()
print("✅ 資料庫與初始資料建立完成。")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def show_tables():
    conn = sqlite3.connect('coffee_shop.db')
    try:
        print("--- 【 menu 菜單表】 ---")
        df_menu = pd.read_sql_query("SELECT * FROM menu", conn)
        display(gr.DataFrame(df_menu) if 'gr' in globals() else df_menu)

        print("--- 【 orders 訂單表】 ---")
        df_orders = pd.read_sql_query("SELECT * FROM orders", conn)
        display(gr.DataFrame(df_orders) if 'gr' in globals() else df_orders)
    except Exception as e:
        print("顯示資料表錯誤：", e)
    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def ai_parse(user_input):
    prompt = f"""
你是一個校園咖啡廳的語音點餐系統助手。
請從使用者的語句話中解析出：動作、品項、數量。
合法動作僅限：訂購、取消、查詢、統計、未知
合法品項僅限：美式咖啡、拿鐵、抹茶歐蕾、未知
數量規則：請從字串中找出整數數量（例如「兩杯」代表 2）。如果沒說數量則預設為 1。
特別注意：如果使用者只說了商品名稱（例如「拿鐵」），動作請自動判斷為「訂購」。如果說「今天各商品賣得怎樣」，動作請判斷為「統計」，品項填「未知」。

請嚴格僅回傳三個欄位，並以逗號分隔，不要有其他字元：
動作,品項,數量

使用者說的話：{user_input}
"""
    raw_response = "未知,未知,1"
    try:
        completion = client.chat.completions.create(
            model=MODEL,
            messages=[{"role": "user", "content": prompt}],
            temperature=0,
            reasoning_effort=REASONING_EFFORT
        )
        raw_response = completion.choices[0].message.content.strip()
    except Exception as e:
        print("⚠️ 呼叫 Groq API 超過額度或發生錯誤 (429/其他)：", e)
        return "未知", "未知", 1, str(e)

    # 解析回傳結果
    parts = [p.strip() for p in raw_response.split(',')]
    action = parts[0] if len(parts) > 0 else "未知"
    item = parts[1] if len(parts) > 1 else "未知"

    # 用正規表達式從文字或結果中找出數字作為數量
    qty_match = re.search(r'\d+', parts[2] if len(parts) > 2 else "1")
    qty = int(qty_match.group()) if qty_match else 1

    # 檢查合法性
    valid_actions = ["訂購", "取消", "查詢", "統計"]
    valid_items = ["美式咖啡", "拿鐵", "抹茶歐蕾"]

    if action not in valid_actions:
        action = "未知"

    # 如果動作是統計，品項允許為未知
    if action == "統計":
        item = "未知"
    elif item not in valid_items:
        item = "未知"

    return action, item, qty, raw_response
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
test_cases = [
    # (使用者說的話, 預期動作, 預期品項, 預期數量) -> "*" 代表不在意
    ("我要兩杯拿鐵", "訂購", "拿鐵", 2),
    ("訂一杯美式咖啡", "訂購", "美式咖啡", 1),
    ("取消拿鐵", "取消", "拿鐵", 1),
    ("查一下抹茶歐蕾", "查詢", "抹茶歐蕾", 1),
    ("今天各商品賣得怎樣", "統計", "未知", 1),
    ("給我十杯拿鐵", "訂購", "拿鐵", 10),
    ("我要珍奶", "訂購", "未知", 1),
    ("拿鐵", "訂購", "拿鐵", 1)
]

print("=== AI 解讀測試 (規範 7-1) ===")
passed_count = 0
for idx, (text, exp_action, exp_item, exp_qty) in enumerate(test_cases, 1):
    action, item, qty, raw = ai_parse(text)
    # 檢查是否符合預期（支援 "*" 不在意比對）
    match_action = (exp_action == "*" or action == exp_action)
    match_item = (exp_item == "*" or item == exp_item)
    match_qty = (exp_qty == "*" or qty == exp_qty)
    is_passed = match_action and match_item and match_qty
    if is_passed:
        passed_count += 1

    status_icon = "✅ 通過" if is_passed else "❌ 失敗"
    print(f"測試 {idx:02d} | 輸入：{text:<12} => 系統解讀：動作={action}, 品項={item}, 數量={qty} | AI原始回覆：{raw} | {status_icon}")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def process_order(name, user_input):
    if not name or not name.strip():
        return "❌ 錯誤：訂購與取消時請務必輸入姓名！"

    name = name.strip()
    action, item, qty, raw = ai_parse(user_input)

    conn = sqlite3.connect('coffee_shop.db')
    cursor = conn.cursor()
    try:
        # 1. 統計查詢
        if action == "統計":
            sql = """
                SELECT
                    m.item AS 品項,
                    m.price AS 單價,
                    m.stock AS 目前庫存,
                    COALESCE(SUM(CASE WHEN o.status = '已訂購' THEN o.quantity ELSE 0 END), 0) AS 已售杯數,
                    m.stock AS 剩餘庫存,
                    (COALESCE(SUM(CASE WHEN o.status = '已訂購' THEN o.quantity ELSE 0 END), 0) * m.price) AS 收入
                FROM menu m
                LEFT JOIN orders o ON m.item = o.item
                GROUP BY m.item, m.price, m.stock
            """
            cursor.execute(sql)
            rows = cursor.fetchall()
            result_str = "📊 【今日銷售統計報告】\n"
            for r in rows:
                result_str += f"- {r[0]} | 單價：{r[1]} | 庫存：{r[2]} | 已售：{r[3]}杯 | 收入：{r[5]}元\n"
            return f"【AI 解析】動作：{action}, 品項：{item}, 數量：{qty}\n\n{result_str}"

        # 2. 查詢庫存
        if action == "查詢":
            if item == "未知":
                return f"❌ 找不到您想查詢的品項。目前品項有：美式咖啡、拿鐵、抹茶歐蕾。"
            cursor.execute("SELECT stock, price FROM menu WHERE item = ?", (item,))
            row = cursor.fetchone()
            if row is None:
                return f"❌ 找不到品項「{item}」。"
            return f"【AI 解析】查詢成功\n💬 系統回覆：{item} 單價 {row[1]} 元，今日剩餘庫存 {row[0]} 杯。"

        # 3. 訂購功能
        if action == "訂購":
            if item == "未知":
                return f"❌ 訂購失敗：找不到您指定的品項。目前品項有：美式咖啡、拿鐵、抹茶歐蕾。"
            if qty < 1 or qty > 3:
                return f"❌ 訂購失敗：數量必須在 1 至 3 杯之間（您輸入了 {qty} 杯）。"

            # 檢查個人今日已訂購總數是否超過 3 杯
            cursor.execute("SELECT SUM(quantity) FROM orders WHERE name = ? AND status = '已訂購'", (name,))
            user_total_row = cursor.fetchone()
            user_current_total = user_total_row[0] if user_total_row and user_total_row[0] is not None else 0

            if user_current_total + qty > 3:
                return f"❌ 訂購失敗：每人每天最多限購 3 杯！您目前已訂購 {user_current_total} 杯，加上本筆會超過上限。"

            # 檢查庫存是否充足
            cursor.execute("SELECT stock, price FROM menu WHERE item = ?", (item,))
            menu_row = cursor.fetchone()
            if menu_row is None:
                return f"❌ 找不到品項「{item}」。"

            stock, price = menu_row[0], menu_row[1]
            if stock < qty:
                return f"❌ 訂購失敗：「{item}」庫存不足！目前僅剩 {stock} 杯。"

            # 扣除庫存並新增訂單
            cursor.execute("UPDATE menu SET stock = stock - ? WHERE item = ?", (qty, item))
            cursor.execute("INSERT INTO orders (name, item, quantity, status) VALUES (?, ?, ?, '已訂購')", (name, item, qty))
            conn.commit()

            total_price = price * qty
            cursor.execute("SELECT stock FROM menu WHERE item = ?", (item,))
            new_stock = cursor.fetchone()[0]

            return f"✅ 【訂購成功】\n👤 姓名：{name}\n☕ 品項：{qty}杯 {item}\n💰 總金額：{total_price} 元\n📦 該品項剩餘庫存：{new_stock} 杯"

        # 4. 取消功能
        if action == "取消":
            if item == "未知":
                return f"❌ 取消失敗：找不到您指定的品項。"
            cursor.execute("SELECT order_id, quantity FROM orders WHERE name = ? AND item = ? AND status = '已訂購' ORDER BY order_id DESC LIMIT 1", (name, item))
            order_row = cursor.fetchone()
            if order_row is None:
                return f"❌ 取消失敗：找不到您（{name}）所訂購的「{item}」有效訂單可供取消。"

            order_id, order_qty = order_row[0], order_row[1]

            cursor.execute("UPDATE orders SET status = '已取消' WHERE order_id = ?", (order_id,))
            cursor.execute("UPDATE menu SET stock = stock + ? WHERE item = ?", (order_qty, item))
            conn.commit()

            cursor.execute("SELECT stock FROM menu WHERE item = ?", (item,))
            restored_stock = cursor.fetchone()[0]

            return f"✅ 【取消成功】\n👤 姓名：{name}\n☕ 已取消 {order_qty} 杯 {item}\n📦 已釋放庫存，目前該品項剩餘：{restored_stock} 杯"

        return f"❓ 無法辨識您的需求（AI 解析：動作={action}, 品項={item}）。"

    except Exception as e:
        conn.rollback()
        return f"⚠️ 執行過程發生錯誤：{e}"
    finally:
        conn.close()
````
